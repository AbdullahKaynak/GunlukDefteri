```python
import pandas as pd
import numpy as np 
import requests
from bs4 import BeautifulSoup
```

### HTML tablolarından veri kazıyın

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/HTMLColorCodes.html"
```

#### Web sayfasının içeriğini metin formatında alın ve 'data' adlı bir değişkende saklayın.

```python
data = requests.get(url).text
soup = BeautifulSoup(data, "html.parser")
```

### Web sayfasındaki bir HTML tablosunu bulun.

```python
table = soup.find("table")
```

### Tablodaki tüm satırları al

```python
for row in table.find_all("tr"):
    # Her satırdaki tüm sütunları alın.
    cols = row.find_all("td")    ## HTML'de bir sütun <td> etiketiyle temsil edilir.
    color_name = cols[2].string  ## 3.sütundaki değeri color_name olarak sakla
    color_code = cols[3].string  ## 4.sütündaki renk kodunu al
    print("{} ----> {}". format(color_name,color_code))
```
