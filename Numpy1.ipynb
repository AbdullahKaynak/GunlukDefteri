# Numpy array dizisi oluşturma

```python
import numpy as np
```

```python
## Liste oluşturma

a = np.array([0, 1, 2, 3, 4])
a
```

```python
# Listelerde olduğu gibi, her bir öğeye köşeli parantez yoluyla erişebiliriz:

print("a[0]:", a[0])
print("a[1]:", a[1])
print("a[2]:", a[2])
print("a[3]:", a[3])
print("a[4]:", a[4])
```

# Değer atama
Dizinin değerini değiştirebiliriz. <code>c</code> dizisini ele alalım:

```python
c = np.array([20, 1, 2, 3, 4 ])
c

## elemöanlarını bu şekilde değiştirebilriiz.
c[0] = 100
c[4] = 0
```

```python
ap = np.array([10, 2, 30, 40, 50])
ap[1] = 20
ap
```

## Slicing (dilimleme)
<code>[başlangıç:bitiş:adım] <code>

```python
# d değişkenine c'nin 1'den 3.değerine kadar ata

d = c[1:4]
d
```

```python
## Yeni değerlere karşılık gelen indeksleri aşağıdaki gibi atayabiliriz:
c[3:5] = 300, 400
c
```

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5:2])
```

```python
arrr = np.array([1, 2, 3, 4, 5, 6, 7, 8])

arrr[1::2]
```

#  Hadamard ürünü
### Aynı konumdaki sayılar birbiriyle çarpılır.
### Matrislerin boyutları birbiriyle tam olarak aynı olmalıdır

```python
## Yol 1 

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

sonuc = a * b
sonuc

## Yol 2 

sonuc2 = np.multiply(a,b)
sonuc2
```

# Nokta Ürünü

```python
A = np.array([[1, 2], 
              [3, 4]])

B = np.array([[5, 6], 
              [7, 8]])

# Nokta Çarpımı (Matris Çarpımı)

sonuc_dot = np.dot(A, B)
# [ [1*5 + 2*7,  1*6 + 2*8],
#   [3*5 + 4*7,  3*6 + 4*8] ]

sonuc_dot
```
