```python
filename = "https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%205/Labs/Golden_State.pkl"

def dowland(url, filename):                 ## dışarıdan bir url ve kaydedilecek dosya adı filename al
    istek = requests.get(url)               ## url'yi getir
    if istek.status_code ==200:             ## başarılı çekersen içeri gir
        with open(filename, "wb") as f:     ## yeni dosya aç bozmadan yaz işi bitince kapa
            f.write(istek.content)

# dowland(filename, "GoldenState.pkl")      ## başarılı indi belge

df = pd.read_pickle("GoldenState.pkl")
df.head()
```
