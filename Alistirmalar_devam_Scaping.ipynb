```python
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
```

```python
html = """
<!DOCTYPE html>
<html>
<head>
<title>Page Title</title>
</head>
<body>
<h3><b id='boldest'>Lebron James</b></h3>
<p> Salary: $ 92,000,000 </p>
<h3> Stephen Curry</h3>
<p> Salary: $85,000, 000 </p>
<h3> Kevin Durant </h3>
<p> Salary: $73,200, 000</p>
</body>
</html>
"""
```

```python
soup = BeautifulSoup(html, "html.parser")
soup
```

```python
print("1- İç içe geçmiş yapısını düzeltelim: \n",soup.prettify())      
```

```python
# Diyelim ki sayfanın başlığını ve en yüksek ücreti alan oyuncunun adını istiyoruz; bu durumda Tag'i kullanabiliriz. 
# Tag nesnesi, orijinal belgedeki bir HTML etiketine (örneğin title etiketine) karşılık gelir.

tag_object = soup.title
print("1- Tittle'ı verelim: ", tag_object)

## `bs4.element.Tag` etiket türünü görebiliriz.
print("2- Etiket türü: ", type(tag_object))

## Aynı ada sahip birden fazla Etiket (Tag) varsa, o Etiket adına sahip ilk öğe çağrılır; 
# bu da en yüksek ücreti alan oyuncuya karşılık gelir:

tag_object = soup.h3
print("3- İlk etiket öğresi: ", tag_object)

## Kalın (bold) `b` özniteliği içine alındığında, ağaç gösterimini kullanmak yararlıdır. 
# `name` değerine ulaşmak için `child` özniteliğini kullanarak ağaç üzerinde aşağı doğru ilerleyebiliriz.

```

```python
tag_child = tag_object.b
print("1- bu kalın yazılmış ilk öge: ", tag_child)

parent_tag = tag_child.parent
print("2-Bu da bunun bir üst ögesini getirir", parent_tag)

print("**********************")

ust_oge = tag_object.parent
print("3-Parent_tag üstü öge:\n ", ust_oge)

print("**********************")

sibling_1 = tag_object.find_next_sibling()
print("4- tag_object sonrası: ", sibling_1)

print("***********************")

sibling_2 = sibling_1.find_next_sibling()
print("5- sibling_1 den sonra olan gelir: ", sibling_2)



```

##### Eğer etiketin öznitelikleri varsa —örneğin `id="boldest"` etiketi, değeri `boldest` olan bir `id` özniteliğine sahiptir— etikete bir sözlük gibi davranarak özniteliklerine erişebilirsiniz:

```python
print(tag_child["id"])
print(tag_child.attrs)
```

##### Python'daki `get()` yöntemini kullanarak etiketin öznitelik içeriğini de elde edebiliriz.

```python
print(tag_child.get("id"))
```

##### Bir string (dizge), bir etiket (tag) içindeki metin veya içeriğe karşılık gelir. Beautiful Soup, bu metni barındırmak için NavigableString sınıfını kullanır. HTML'imizde, `tag_child` adlı Tag nesnesinin string değerini şu şekilde çıkararak ilk oyuncunun adını elde edebiliriz:

```python
tag_string = tag_child.string
print(tag_string)
print(type(tag_string))
```

```python
html = """
<table>
  <tr>
    <td id='flight' >Flight No</td>
    <td>Launch site</td> 
    <td>Payload mass</td>
   </tr>
  <tr> 
    <td>1</td>
    <td><a href='https://en.wikipedia.org/wiki/Florida'>Florida</a></td>
    <td>300 kg</td>
  </tr>
  <tr>
    <td>2</td>
    <td><a href='https://en.wikipedia.org/wiki/Texas'>Texas</a></td>
    <td>94 kg</td>
  </tr>
  <tr>
    <td>3</td>
    <td><a href='https://en.wikipedia.org/wiki/Florida'>Florida</a> </td>
    <td>80 kg</td>
  </tr>
</table>
"""
```

```python
table = html
table_bs = BeautifulSoup(table, "html.parser")
```

##### `name` parametresini bir etiket adına ayarladığımızda, yöntem o ada sahip tüm etiketleri ve bunların alt öğelerini ayıklayacaktır.

```python
table_rows = table_bs.find_all("tr")
print("1- Bütün dosyayı yaz: \n",table_rows)

print("**********************************")

first_row = table_rows[0]
print("2- tablonun ilkini al: ",first_row)

print("**********************************")

print(first_row.td)

print("**********************************")

for i,row in enumerate(table_rows):
    print("row",i,"is",row)

print("**********************************")

```

```python
list_input = table_bs.find_all(name =["tr", "td"])
print("tr ve td listesi: ",list_input)
```

```python
list_input=table_bs.find_all(href="https://en.wikipedia.org/wiki/Florida")
list_input
```

```python
html3 = """<h3>Rocket Launch </h3>

<p>
<table class='rocket'>
  <tr>
    <td>Flight No</td>
    <td>Launch site</td> 
    <td>Payload mass</td>
  </tr>
  <tr>
    <td>1</td>
    <td>Florida</td>
    <td>300 kg</td>
  </tr>
  <tr>
    <td>2</td>
    <td>Texas</td>
    <td>94 kg</td>
  </tr>
  <tr>
    <td>3</td>
    <td>Florida </td>
    <td>80 kg</td>
  </tr>
</table>
</p>
<p>

<h3>Pizza Party  </h3>
  
    
<table class='pizza'>
  <tr>
    <td>Pizza Place</td>
    <td>Orders</td> 
    <td>Slices </td>
   </tr>
  <tr>
    <td>Domino's Pizza</td>
    <td>10</td>
    <td>100</td>
  </tr>
  <tr>
    <td>Little Caesars</td>
    <td>12</td>
    <td >144 </td>
  </tr>
  <tr>
    <td>Papa John's </td>
    <td>15 </td>
    <td>165</td>
  </tr>"""
```

##### BeautifulSoup'tan two_tables_bs adında bir nesne oluşturuyoruz.

```python
Two_tables = BeautifulSoup(html3, "html.parser")
```

İlk tabloyu `table` etiket adını kullanarak bulabiliriz.

```python
first_table = Two_tables.find("table")
first_table
```

İkinci tabloyu bulmak için class özniteliğine göre filtreleme yapabiliriz; ancak class Python'da bir anahtar sözcük olduğu için bir alt çizgi ekleriz.

```python
Two_tables.find("table", class_="pizza")
```
