```python
import requests
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
```

```python
from IPython.core.display import HTML

html_code = """
<!DOCTYPE html>
<html>
<head>
<title> Page Title </title>
</head>
<body>
<h3><b id="boldest"> Lebron James </b></h3>
<p> Salary : $ 92,000,000 </p>
<h3> Stephen Curry </h3>
<p>Salary : $ 85,000,000</p>
<h3> Kevin Durant </h3>
<p>Salary : $ 73,200,000</p>
</body>
</html>
"""
HTML(html_code)
```

```python
soup = BeautifulSoup(html_code, "html5lib")
print("1 : ",soup.title)            # 1
print("2 : ",soup.h3)           # 2
print("3 : ",soup.b)            # 3

tag_child = soup.b
print("4 :", tag_child)         # 4

parent_tag = tag_child.parent
print("5 :",parent_tag)         #5

sibling_1 = parent_tag.find_next_sibling()
print("6 :", sibling_1)         #6

print("7: " ,tag_child.attrs)   # 7 öznitelik adı ve değeri sözlük olarak
print("8 :", tag_child.string)  # 8 içeriği de gezebilirsin string ifadesi için

```

```python
html_content ="""
<table>
  <tr>
    <td>Pizza Place</td>
    <td>Orders</td>
    <td>Slices</td>
  </tr>
  <tr>
    <td>Domino's Pizza</td>
    <td>10</td>
    <td>100</td>
  </tr>
  <tr>
    <td>Little Caesars</td>
    <td>12</td>
    <td>144</td>
  </tr>
</table>
"""
```

```python
table = BeautifulSoup(html_content, "html5lib")

table_row = table.find_all(name= "tr")          ## tr sütununu verdi
print("1- Bütün kolon isimleri: \n",table_row)

print("****************")

first_row = table_row[0]                        ## ilk sütunumuzu çekebiliriz tablodaki
print("2- Tablodaki ilk row: \n", first_row)

print("****************")

first_row_cell = first_row.td
print("3-İlk tablonun ilk hücresi: ", first_row_cell)

print("*****************")

for i,row in enumerate(table_row):
    print("row", i)
    cells = row.find_all("td")                  ## iç içe geçmiş bir for döngüsü ile tabloyu tek tek açıklayıcı dönelim

    for j,cell in enumerate(cells):
        print("column", j, "cell", cell)
```
