## Exercise 3

```python
## Kullanıcılarla ilgili bilgileri içeren bir tablo oluşturmak için, istenen 
#  tüm parametreleri kapsayan bir fonksiyon yazabiliriz; örneğin isim, cinsiyet, şehir vb.

#"Get" metotlarını çağırır ve ardından kullanıcıları içeren bir pandas DataFrame'i döndürür.

from psutil import users


def get_users():
    users =[]

    for user in RandomUser.generate_users(10):
        users.append({"name": user.get_full_name(), "Gender": user.get_gender(), "City": user.get_city,
                      "State": user.get_state(), "Email": user.get_email(), "Dob":user.get_dob(), "Picture": user.get_picture() })

    return pd.DataFrame(users)
```

## Excersize 3 (Fruityvice API)

```python
import requests
import json
```

```python
data = requests.get("https://web.archive.org/web/20240929211114/https://fruityvice.com/api/fruit/all", timeout=30)      ## 30 saniye sürsün alma
data.raise_for_status()

# Sonuçları `json.loads()` fonksiyonunu kullanarak alacağız.
result = json.loads(data.text)

## burayı bir datafram'e dönüştür
df = pd.DataFrame(result)
df.head()
```

```python
# kirazın familyasını ve cinsini bilmemiz gerekiyor.
cherry = df.loc[df["name"] == "Cherry"]
print(cherry.iloc[0]["family"]) 
print(cherry.iloc[0]["genus"])
```

```python
## bir muzun kaç kalori içerdiğini öğrenin.
banana_calories = df.loc[df["name"] == "Banana", "nutritions"].iloc[0]["calories"]
print(banana_calories)
```
